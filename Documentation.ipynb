## Introduction
A Neural Network is a type of self learning artificial intelligence that tries to simulate the brain. In my example, I'm making a very simple neural network to identify a hand drawn digit from 0-9. Before we delve into how AI would do this, lets first think how we humans know which digit a drawing is. First, to get these digits, I will be making use of the MNIST dataset of handwritten digits, kindly compiled by the researchers at the National Institute of Standards and Technology, USA. An example snippet of these has been given below.

We may look at a single digit, and instantly recognise that digit. Its not something difficult for a human to do. But take the 5 instances of a 1 below.

<br/>
Each 1 looks different, they are distinctly different images, they induce different cells in your eyes to send diferent signals to your brain but your brain can classify each of those drawings into the same digit 1. So there is a pattern in all those 5 images above which your brain can use to classify those images, and we can try to train a machine to spot the same patterns for the same goal. But this pattern matching, you can do it in an instant, but its practically impossiblle to figure out how our brains actually spot these patterns. We cant just write a computer program with distinct rules to spot the digit, because we dont even know what rules to write, how the brain does this. So, what if instead of telling a computer how exactly to recognise a digit, we just instead write a program for the computer to run that makes it sort of simulate the brain, then it can figure out the patterns that define each digit by itself. Thats the main idea behind neural networks, as the name suggests, to sort of attempt to simulate the brain and the way the brain works so that the computer can artificially think to solve a problem (hence AI, Artificial Intelligence) and artificially learn to get better at solving it (hence ML, Machine Learning). All we do to teach the computer is to just hand it a bunch of these sort of images, let it predict what the digit actually is, and then correct it, and let it adjust itself and learn. if we repeat that enough, then the computer will adjust itself to successfully be able to classify digits just like a human.
